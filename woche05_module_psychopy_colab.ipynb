## 1. Was ist ein Modul?

Ein **Modul** ist eine Python-Datei (`.py`) mit Funktionen, Variablen und Klassen – bereit zum Importieren.

Python's Standardbibliothek enthält Module, die ohne Installation verfügbar sind:

| Modul | Inhalt | Experimenteller Einsatz |
|---|---|---|
| `random` | Zufallszahlen, Mischen | Stimuli randomisieren |
| `pathlib` | Dateipfade, Ordner | Logfiles anlegen |
| `csv` | CSV-Dateien | Ergebnisse speichern |
| `time` | Zeitmessung | Grobe Zeitnahme |

### 3.2 Reproduzierbarkeit mit `seed()`

Pseudozufällig: Eine Startzahl bestimmt die gesamte Folge. In der Wissenschaft unverzichtbar.

````python
import random
````

````python
# Mit seed: immer dieselbe Reihenfolge
myseed = 42
random.seed(myseed)
stimuli = ["ROT", "BLAU", "GRÜN", "GELB"]
random.shuffle(stimuli)
print("Mit seed:", myseed, stimuli)
````

````python
# Nochmal mit einem anderen seed:
myseed = 45

random.seed(myseed)
stimuli2 = ["ROT", "BLAU", "GRÜN", "GELB"]
random.shuffle(stimuli2)
print("Nochmal mit seed ", myseed, stimuli2)
````

````python
# Ohne seed: jedes Mal anders
stimuli3 = ["ROT", "BLAU", "GRÜN", "GELB"]
random.shuffle(stimuli3)
print("Ohne seed:  ", stimuli3)
````

### 3.3 Trialliste mit `zip()` randomisieren

`zip()` verbindet zwei Listen elementweise: `zip(["a","b"], [1,2])` → `[("a",1), ("b",2)]`

````python
import random

SEED = 42

woerter = ["ROT", "BLAU", "GRÜN"] * 4    # je 4 Wiederholungen
farben  = ["rot", "blau", "grün"] * 4

trials = list(zip(woerter, farben))
print(f"Vor Randomisierung:  {trials[:4]} ...")

random.seed(SEED)
random.shuffle(trials)
print(f"Nach Randomisierung: {trials[:4]} ...")
````

````python
# Wie sieht das Ganze ohne seed aus?
import random

woerter2 = ["ROT", "BLAU", "GRÜN"] * 4    # je 4 Wiederholungen
farben2  = ["rot", "blau", "grün"] * 4

trials = list(zip(woerter2, farben2))
print(f"Vor Randomisierung:  {trials[:4]} ...")

random.shuffle(trials)
print(f"Nach Randomisierung: {trials[:4]} ...")
````

````python
import random

woerter = ["ROT", "BLAU", "GRÜN"] * 4
farben  = ["rot", "blau", "grün"] * 4

random.shuffle(woerter)

my_trials = list(zip(woerter, farben))
random.shuffle(my_trials)

# Bedingung hinzufügen
trials_mit_bedingung = []

for wort, farbe in my_trials:
    if wort.lower() == farbe:
        bedingung = "kongruent"
    else:
        bedingung = "inkongruent"
    trials_mit_bedingung.append((wort, farbe, bedingung))

trials_mit_bedingung
````

## 5. `csv` – Logfiles schreiben

CSV (Comma-Separated Values) ist das Standardformat für Psychologie-Daten – lesbar von SPSS, R und Excel.

````python
import csv
from pathlib import Path

# Ordner anlegen
Path("daten").mkdir(exist_ok=True)
logfile = Path("daten") / "vp01_stroop.csv"

# Datei anlegen + Kopfzeile schreiben (mode="w" = neu erstellen)
with open(logfile, mode="w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["trial", "wort", "farbe", "antwort", "korrekt", "rt_ms"])

print(f"Logfile angelegt: {logfile}")
````

## 7. Übergang zu PsychoPy

### 7.1 Warum PsychoPy und nicht reines Python?

| Problem | Reines Python | PsychoPy |
|---|---|---|
| Zeitgenauigkeit | `time.sleep()` ±10–50 ms ungenau | Sub-Millisekunden-Präzision |
| Darstellung | Kein Grafikfenster | Vollbild mit präzisem Timing |
| Tastatureingabe | `input()` blockierend | Nicht-blockierendes Event-Polling |
| Stimuli | Nur Konsolentext | Text, Bilder, Töne, Gitter |

PsychoPy läuft **lokal auf Ihrem Computer**, nicht in Colab, d.h. es braucht ein Grafikfenster.

```bash
# Installation im Terminal:
pip install psychopy
```

### 7.2 Die Grundbausteine

```
1. Imports        → from psychopy import visual, core, event
2. Konfiguration  → Konstanten, VP-ID, Pfade
3. Setup          → Fenster öffnen, Uhr erstellen
4. Stimulus-Setup → TextStim, ImageStim vorbereiten
5. Experiment     → Trial-Schleife: zeigen → warten → speichern
6. Abschluss      → fenster.close(), core.quit()
```

### 7.3 Was Sie wiedererkennen

| PsychoPy | Woher bekannt |
|---|---|
| `VP_ID = "VP01"` | Woche 1: Variablen |
| `trials = [("ROT","blue"), ...]` | Woche 2: Listen/Tupel |
| `for wort, farbe in trials:` | Woche 3: Schleifen |
| `def zeige_trial(wort, farbe):` | Woche 4: Funktionen |
| `from psychopy import visual` | **Woche 5: Module** |

## Zusammenfassung

| Konzept | Syntax | Einsatz |
|---|---|---|
| Import | `import random` | Modul einbinden |
| Selektiv | `from pathlib import Path` | Einzelne Klassen |
| Shuffle | `random.shuffle(liste)` | Trialreihenfolge |
| Seed | `random.seed(42)` | Reproduzierbarkeit |
| Pfad | `Path("daten") / "log.csv"` | OS-unabhängig |
| Logfile | `csv.writer` + `mode="w"` | Kopfzeile |
| Anfügen | `csv.writer` + `mode="a"` | Trials speichern |
| Sicher öffnen | `with open(...) as f:` | Datei immer geschlossen |

**Drei Regeln:**
1. `random.seed()` **einmal** am Anfang
2. Pfade immer mit `pathlib.Path`
3. Dateien immer mit `with open(...) as f:`
